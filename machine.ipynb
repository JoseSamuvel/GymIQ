```python
import pandas as pd
import numpy as np

# 1. Load the raw dataset
df = pd.read_csv('gym_members_dataset.csv')
print(f"Original Dataset Shape: {df.shape}")

# 2. Drop personal identifier columns (not useful for machine learning)
cols_to_remove = ['Member_ID', 'Name', 'Address', 'Phone_Number']
df_cleaned = df.drop(columns=cols_to_remove)

# 3. Handle missing values in numerical columns using Median imputation
numerical_cols = ['Age', 'Avg_Calories_Burned', 'Total_Weight_Lifted_kg', 'Visits_Per_Month']
for col in numerical_cols:
    median_val = df_cleaned[col].median()
    df_cleaned[col] = df_cleaned[col].fillna(median_val)

# 4. Feature Engineering from Dates (Compute Tenure in Days)
df_cleaned['Join_Date'] = pd.to_datetime(df_cleaned['Join_Date'])
df_cleaned['Last_Visit_Date'] = pd.to_datetime(df_cleaned['Last_Visit_Date'])

# Calculate tenure in days between join date and last visit
df_cleaned['Tenure_Days'] = (df_cleaned['Last_Visit_Date'] - df_cleaned['Join_Date']).dt.days

# Fill missing tenure with the median tenure
median_tenure = df_cleaned['Tenure_Days'].median()
df_cleaned['Tenure_Days'] = df_cleaned['Tenure_Days'].fillna(median_tenure)

# Drop the raw date columns after extraction
df_cleaned = df_cleaned.drop(columns=['Join_Date', 'Last_Visit_Date'])

# 5. Encode Target Column ('Churn': 'Yes' -> 1, 'No' -> 0)
df_cleaned['Churn'] = df_cleaned['Churn'].map({'Yes': 1, 'No': 0})

# 6. One-Hot Encode Categorical Features
categorical_cols = ['Gender', 'Membership_Type', 'Favorite_Exercise']
df_final = pd.get_dummies(df_cleaned, columns=categorical_cols, drop_first=True, dtype=int)

# 7. Verify Data Cleaning
print("\nMissing values check:")
print(df_final.isnull().sum())
print(f"\nCleaned Dataset Shape: {df_final.shape}")

# 8. Save the cleaned dataset to CSV
df_final.to_csv('cleaned_gym_members.csv', index=False)
print("\nCleaned dataset successfully saved to 'cleaned_gym_members.csv'!")
```
